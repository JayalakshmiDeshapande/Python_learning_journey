```python
class Human:
    def __init__(self, name="unknown", age=0, salary=-1):#optional parameters
        print("constuctor is called", name)
        self.name = name
        self.age = age
        self.salary=salary

    def walk(self):
        print(self.name, "is walking")

    def work(self):
        print(self.name, "is working")

    def sleep(self):
        print(self.name, "is sleeping")
j=Human("jayu",22,50000)
j.walk()
j.work()
j.sleep()
j=Human("shri",23)
j.walk()
j.work()
j.sleep()
someone=Human()
someone.walk()
someone.work()
someone.sleep()

class Movie:
    def __init__(self,title="idk",rating=3):
      self.title=title
      self.rating=rating

    def display_title(self):
        print(self.title,"is the movies title")
    def display_rating(self):
        print(self.rating,"is the movies rating")
m=Movie("kantara", 5)
print(m.title)
print(m.rating)
m.display_title()
m.display_rating()

m=Movie()
print(m.title)
print(m.rating)
m.display_title()
m.display_rating()
```
