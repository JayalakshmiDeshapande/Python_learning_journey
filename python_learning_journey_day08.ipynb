```python
#all about functions
def greet():
  print("hello, good morning")
greet()

def info(name,country):#parameters
  print(f"{name} from {country}")
info("jayu","India")#positional arguments
info(country="USA",name="john")#keyword arguments
print(info)

#setting default parameters
def marriage(girl,boy="engineering"):
  print(f"{girl} is married to {boy}")
marriage("shreya")
marriage("shri")

#using return
def funct(num):
  return int(str(num)*2)
a=10
b=funct(2)
c=a+b
print(c)

a=5
c=a+funct(3)
print(c)

#local variable and global variable
y = "shri" # Define y as a global variable before calling funct()

def funct_scope_example():
  x=10 # x is a local variable to funct_scope_example
  print("Inside function: x (local) =", x)
  print("Inside function: y (global) =", y)

funct_scope_example()
print("Outside function: y (global) =", y)
# The following line would cause a NameError because x is local to funct_scope_example()
# print("Outside function: x (local) =", x)
```
