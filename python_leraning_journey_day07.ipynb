```python
l=[1,23,43,534,32]
dl=[]
for num in l:
  dl.append(num*2)
  print(dl)


student_marks={
    "Anand": 85,
    "Geetha":90,
    "kumar":90
}
for student, marks in student_marks.items():
  print(f"{student}:{marks}")

#adding list to dictionary
student=["Chandan","Darshan","Narendra"]
marks=[25,99,78]
student_marks={}
for i in range(len(student)):
  student_marks[student[i]]=marks[i]
  print(student_marks)

#list comprehension(syntax: exp for item in collection if condition)
l=[1,2,3,4,5]
dl=[item*2 for item in l if item%2==0]
print(dl)

#dictionary comprehension
student=["Chandan","Darshan","Narendra"]
marks=[25,99,78]
student_marks={student:marks for student,marks in zip(student,marks)}
print(student_marks)
'''zip() combines different lists together, while enumerate()
 adds an index to items within a single list.'''
s=("This_is_eagle")
l=s.split("_")
print(l)
s=("This is eagle")
print(list(s))

#taking list as a input
x=input("enter list of integers:")
print(x.split())# it prints list as list as we used split().
#suppose if we did not use split(), it returns as string only.
l=[int(num) for num in x.split()]
print(l)
```
