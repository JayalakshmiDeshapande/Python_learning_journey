```python
#setting variable length argument
def add(*a): #variable length argument
  total_sum = 0
  for i in a:
    total_sum = total_sum + i
  return total_sum
print(add(1,2,3,4,5))

def student_info(**details):
  for key,value in details.items():
    print(f"{key}:{value}")
student_info(name="jayu", marks=100)
```

### Here's how to define and use a lambda function:

```python
add_ten = lambda x: x + 10
print(add_ten(5))

multiply = lambda a, b: a * b
print(multiply(3, 4))
```

You can also use lambda functions as arguments to higher-order functions like `map()`, `filter()`, and `sorted()`.

```python
# Using lambda with filter()
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even numbers: {even_numbers}")

# Using lambda with map()
squared_numbers = list(map(lambda x: x**2, numbers))
print(f"Squared numbers: {squared_numbers}")

add=lambda x,y:x+y
print(add(2,3))

#Recursion
def factorial(n):
  if n==1:
    return 1
  return n*factorial(n-1)
print(factorial(5))

#nested functions
def calculate(a,b):
  def add():
    return a+b
  def sub():
    return a-b
  return add(), sub()
print(calculate(10,20))
```
